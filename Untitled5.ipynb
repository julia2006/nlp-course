```python
%%writefile app.py
import pandas as pd
import streamlit as st

# Configure page title and layout
st.set_page_config(page_title="SNCF TGV Punctuality Analysis", layout="centered")
st.title("TGV Punctuality Analysis (2018–2026)")


@st.cache_data
def load_data(filepath: str) -> pd.DataFrame:
    df = pd.read_csv(filepath, sep=";")
    df["Year"] = df["Date"].str[:4].astype(int)
    df["Month"] = df["Date"].str[5:7].astype(int)
    df["Circulations réelles"] = (
        df["Nombre de circulations prévues"] - df["Nombre de trains annulés"]
    )
    return df


df = load_data("regularite-mensuelle-tgv-aqst (2).csv")

st.caption(
    f"Dataset: Régularité mensuelle TGV (AQST) — {len(df):,} route-month observations "
    f"across {df['Date'].nunique()} months ({df['Date'].min()} to {df['Date'].max()})"
)

# Optional service filter (National, International, or All)
service_options = ["All"] + sorted(df["Service"].unique().tolist())
selected_service = st.selectbox("Service scope:", service_options)

filtered_df = df if selected_service == "All" else df[df["Service"] == selected_service]

# Compute Jan–Sep 2025 vs. 2026 punctuality dynamically
jan_sep = filtered_df[
    filtered_df["Month"].between(1, 9) & filtered_df["Year"].isin([2025, 2026])
]

agg = jan_sep.groupby("Year").agg(
    circ_reelles=("Circulations réelles", "sum"),
    retards_arrivee=("Nombre de trains en retard à l'arrivée", "sum"),
    retards_depart=("Nombre de trains en retard au départ", "sum"),
)

agg["Arrival regularity"] = (1 - agg["retards_arrivee"] / agg["circ_reelles"]) * 100
agg["Departure punctuality"] = (1 - agg["retards_depart"] / agg["circ_reelles"]) * 100

comparison_data = pd.DataFrame({
    "Indicator": ["Arrival regularity", "Departure punctuality"],
    "2025 (Jan–Sep)": [
        round(agg.loc[2025, "Arrival regularity"], 2),
        round(agg.loc[2025, "Departure punctuality"], 2),
    ],
    "2026 (Jan–Sep)": [
        round(agg.loc[2026, "Arrival regularity"], 2),
        round(agg.loc[2026, "Departure punctuality"], 2),
    ],
})
comparison_data["Change (pp)"] = (
    comparison_data["2026 (Jan–Sep)"] - comparison_data["2025 (Jan–Sep)"]
).round(2)

st.subheader("Finding 1: TGV Performance Declined in 2026")
st.dataframe(comparison_data, use_container_width=True)

# Interactive metric selector
selected_indicator = st.selectbox(
    "Select an indicator to highlight:",
    comparison_data["Indicator"],
)

row = comparison_data[comparison_data["Indicator"] == selected_indicator].iloc[0]
st.metric(
    label=f"{selected_indicator} (Jan–Sep 2026 — {selected_service})",
    value=f"{row['2026 (Jan–Sep)']:.2f}%",
    delta=f"{row['Change (pp)']:.2f} pp vs. 2025",
)
```
